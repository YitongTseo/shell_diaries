```python
import nltk
from nltk import word_tokenize
from nltk.probability import FreqDist
import re

# Download required NLTK data (first time only)
nltk.download('punkt')
nltk.download('stopwords')

# Your list of proper nouns to filter out
proper_nouns = [
    'Kino', 'Juana', 'Coyotito', 'Juan', 'Tomas',
    'La', 'Paz', 'Gulf', 'California', 'Mexico',
    # Add more names/places as you find them
]

# Convert to lowercase for matching
proper_nouns_lower = [name.lower() for name in proper_nouns]

# Read the text file
with open('the_pearl.txt', 'r', encoding='utf-8') as f:
    text = f.read()

# Tokenize
tokens = word_tokenize(text.lower())

# Filter out:
# 1. Proper nouns from your list
# 2. Punctuation and non-alphabetic tokens
# 3. Optionally: stopwords (common words like 'the', 'a', 'and')
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens 
    if word.isalpha()  # Only alphabetic words
    and word not in proper_nouns_lower  # Remove proper nouns
    # and word not in stop_words  # Uncomment to remove stopwords too
]

# Calculate frequency distribution
fdist = FreqDist(filtered_tokens)

# Get top N words
top_n = 1024  # Change this to whatever you want
top_words = fdist.most_common(top_n)

# Display results
print(f"Total words analyzed: {len(filtered_tokens)}")
print(f"\nTop {min(top_n, len(top_words))} words by frequency:\n")
for word, freq in top_words[:50]:  # Show first 50
    print(f"{word:20} {freq}")

# Save to file
with open('top_words_from_the_pearl.txt', 'w', encoding='utf-8') as f:
    for word, freq in top_words:
        f.write(f"{word}\t{freq}\n")
```
